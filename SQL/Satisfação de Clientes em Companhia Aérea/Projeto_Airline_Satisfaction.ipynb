```python
#Importando bibliotecas e criando a conexão com o SQLite3
import pandas as pd
from sqlite3 import connect
conn = connect(":memory:")
```

```python
#Lendo os arquivos .csv e concatenando, uma vez que não usaremos técnicas de ML com divisão train e test
df_train = pd.read_csv('/content/sample_data/train.csv')
df_test = pd.read_csv('/content/sample_data/test.csv')
df = pd.concat([df_train, df_test])
```

```python
df.head()
```

```python
#Checando se existem valores nulos nos dados
print('Número de valores total: ', df.shape[0])
print('           Coluna         |   N° de valores nulos\n')
df.isna().sum()
```

```python
#Substituindo os valores nulos nos dados pela mediana da coluna
df['Arrival Delay in Minutes'] = df['Arrival Delay in Minutes'].fillna(df['Arrival Delay in Minutes'].median(axis = 0))
```

```python
#Confirmando que os valores foram substituidos
print('Número de valores total: ', df.shape[0])
print('           Coluna         |   N° de valores nulos\n')
df.isna().sum()
```

```python
#Checando número de linhas duplicadas
print('Número de linhas duplicadas: ', df.duplicated().sum())
```

```python
#Calculando a maior e a menor idade nos dados, para criamos os bins
print('A maior idade nos dados é: ', df['Age'].max())
print('A menor idade nos dados é: ', df['Age'].min())
```

```python
#Criando a coluna de idade com bins (categorias)
df['Age_bin'] = pd.cut(df['Age'], bins=[0, 20, 30, 40, 50, 60, 70, 80, 90], labels = ["0-20", "21-30", "31-40", "41-50", "51-60", "61-70", "71-80", "81-90"])
```

```python
#Conferindo se nossos bins de idade foram criados corretamente
df['Age_bin'].value_counts()
```

```python
#Calculando a maior e a menor distância de viagem nos dados, para criamos os bins
print('A maior distância de viagem nos dados é: ', df['Flight Distance'].max())
print('A menor distância de viagem nos dados é: ', df['Flight Distance'].min())
```

```python
#Criando a coluna de distância de viagem com bins (categorias)
df['Distance_bins'] = pd.cut(df['Flight Distance'], bins=[0, 500, 1500, 3000, 4500, 6000], labels = ["0-500", "501-1500", "1501-3000", "3001-4500", "4501-6000"])
```

```python
#Conferindo se nossos bins de distância foram criados corretamente
df['Distance_bins'].value_counts()
```

```python
#Calculando o maior e o menor atraso na decolagem nos dados, para criamos os bins
print('O maior atraso na decolagem, em minutos, nos dados é: ', df['Departure Delay in Minutes'].max())
print('O menor atraso na decolagem, em minutos, nos dados é: ', df['Departure Delay in Minutes'].min())
```

```python
#Criando a coluna de atraso na decolagem com bins (categorias)
df['Departure_bins'] = pd.cut(df['Departure Delay in Minutes'], bins=[0, 1, 120, 240, 480, 840, 1200, 1800], labels = ["Sem atraso","Até 2 horas", "2-4 horas", "4-8 horas", "8-14 horas", "14-20 horas", "+20 horas"], include_lowest=True)
```

```python
#Conferindo se nossos bins de atraso na decolag foram criados corretamente
df['Distance_bins'].value_counts()
```

```python
#Calculando o maior e o menor atraso na chegada nos dados, para criamos os bins
print('O maior atraso na chegada, em minutos, nos dados é: ', df['Arrival Delay in Minutes'].max())
print('O menor atraso na chegada, em minutos, nos dados é: ', df['Arrival Delay in Minutes'].min())
```

```python
#Criando a coluna de atraso na chegada com bins (categorias)
df['Arrival_bins'] = pd.cut(df['Arrival Delay in Minutes'], bins=[0, 1, 120, 240, 480, 840, 1200, 1800], labels = ["Sem atraso","Até 2 horas", "2-4 horas", "4-8 horas", "8-14 horas", "14-20 horas", "+20 horas"], include_lowest=True)
```

```python
#Conferindo se nossos bins de atraso na chegada foram criados corretamente
df['Arrival_bins'].value_counts()
```

O atraso na decolagem está correlacionado com o atraso na chegada. Se o avião decolou atrasado, ele irá provavelmente aterrisar atrasado. Queremos observar se a satisfação do cliente é alterada se o atraso ocorre somente na chegada, e não na decolagem. Chamaremos esta variável de Atraso de Percurso, uma vez que estamos lidando com atrasos que foram causados durante o voo. Valores positivos nesta variável indicam que houve atraso no percurso. Valores negativos indicam que o percurso foi mais rápido do que o esperado e "compensou" parte do atraso na decolagem.

```python
df['Flight Delay in Minutes'] = df['Arrival Delay in Minutes'] - df['Departure Delay in Minutes']
```

```python
print('O maior atraso de percurso, em minutos, nos dados é: ', df['Flight Delay in Minutes'].max())
print('O menor atraso de percurso, em minutos, nos dados é: ', df['Flight Delay in Minutes'].min())
```

```python
#Convertendo todos as colunas para letras minúsculas e os espaços para _, para facilitar a leitura em SQL
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ', '_')
df.columns = df.columns.str.replace('/', '_')
df.columns = df.columns.str.replace('-', '_')
```

```python
#Convertendo dataframe em SQL
df.to_sql(name="data", con=conn)
pd.read_sql("SELECT * FROM data LIMIT 5", con = conn)
```
